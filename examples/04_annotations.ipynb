# Statistical Annotations and Visual Elements

This notebook demonstrates yplot's annotation tools:
- Significance brackets and p-values
- Arrows and callouts
- Scale bars
- Text boxes

**All examples use SubplotLayout** for precise control over figure dimensions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import yplot
import yplot.plot as yp
from yplot.annotate import (
    add_significance, add_pvalue, star_notation,
    add_arrow, add_callout, add_scale_bar, add_text_box
)
from yplot.annotate.stats import add_significance_bars, add_regression_stats

yplot.use('publication')
```

## Statistical Significance Brackets

```python
# Create bar chart data
conditions = ['Control', 'Drug A', 'Drug B', 'Drug C']
means = [10, 18, 15, 25]
stds = [2, 3, 2.5, 4]

layout = yplot.SubplotLayout(config={
    "fig_size": (6, 5.5),
    "margins": {"left": 0.7, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (5.0, 4.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

bars = ax.bar(range(4), means, yerr=stds, capsize=3)
ax.set_xticks(range(4))
ax.set_xticklabels(conditions)
ax.set_ylabel('Response')
ax.set_ylim(0, 35)

# Add significance brackets
add_significance(ax, 0, 1, 22, pvalue=0.01)   # Control vs Drug A
add_significance(ax, 0, 3, 27, pvalue=0.001)  # Control vs Drug C
add_significance(ax, 1, 3, 32, pvalue=0.05)   # Drug A vs Drug C

ax.set_title('Treatment Comparison with Significance')

plt.show()
```

## Star Notation Helper

```python
# Demonstrate star notation conversion
pvalues = [0.0001, 0.005, 0.03, 0.1]
for p in pvalues:
    print(f"p = {p} -> {star_notation(p)}")
```

## Multiple Comparisons with Automatic Stacking

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (6, 5.5),
    "margins": {"left": 0.7, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (5.0, 4.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

bars = ax.bar(range(4), means, yerr=stds, capsize=3)
ax.set_xticks(range(4))
ax.set_xticklabels(conditions)
ax.set_ylabel('Response')

# Use automatic stacking
comparisons = [
    (0, 1, 0.01),   # Control vs Drug A
    (0, 2, 0.05),   # Control vs Drug B
    (0, 3, 0.001),  # Control vs Drug C
]
add_significance_bars(ax, comparisons, y_start=30)

ax.set_title('Multiple Comparisons')

plt.show()
```

## Regression Statistics

```python
np.random.seed(42)
x = np.linspace(0, 10, 50)
y = 2.5 * x + 3 + np.random.normal(0, 3, 50)

layout = yplot.SubplotLayout(config={
    "fig_size": (5.5, 4.5),
    "margins": {"left": 0.7, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (4.5, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

ax.scatter(x, y, alpha=0.7)

# Fit line
from scipy import stats
slope, intercept, r_value, p_value, _ = stats.linregress(x, y)
ax.plot(x, slope * x + intercept, 'r-', lw=1.5)

# Add regression stats
add_regression_stats(
    ax, 
    r_squared=r_value**2, 
    pvalue=p_value,
    slope=slope,
    intercept=intercept,
    loc='lower right'
)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Scatter with Regression Stats')

plt.show()
```

## Arrows and Callouts

```python
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = np.sin(x) + np.random.normal(0, 0.1, 100)

layout = yplot.SubplotLayout(config={
    "fig_size": (8, 4.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (7.0, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

ax.plot(x, y)

# Add callout to peak
peak_idx = np.argmax(y)
add_callout(
    ax, x[peak_idx], y[peak_idx], 
    'Maximum\nvalue', 
    x[peak_idx] + 1.5, y[peak_idx] + 0.3
)

# Add arrow pointing to feature
add_arrow(
    ax, 7, -0.8, 5.5, -0.5,
    text='Minimum region',
    color='red'
)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Plot with Callouts and Arrows')

plt.show()
```

## Scale Bars

```python
# Simulate microscopy-like data
np.random.seed(42)
img = np.random.rand(100, 100)

layout = yplot.SubplotLayout(config={
    "fig_size": (5.5, 5.5),
    "margins": {"left": 0.2, "right": 0.2, "top": 0.4, "bottom": 0.2},
    "row_1": {
        "size": (5.0, 4.8),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

ax.imshow(img, cmap='gray')
ax.axis('off')

# Add scale bar (e.g., 20 units = 10 um)
add_scale_bar(ax, length=20, unit='um', label='10 um', loc='lower right')

ax.set_title('Image with Scale Bar')

plt.show()
```

## Text Boxes

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (6, 4.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (5.0, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

# Create some data
x = np.linspace(0, 10, 50)
y = x ** 2
ax.plot(x, y)

# Add sample size info
add_text_box(ax, 0.05, 0.95, 'n = 50\nR2 = 0.99', transform='axes')

# Add equation
add_text_box(
    ax, 0.95, 0.5, 'y = x^2', 
    transform='axes', ha='right',
    facecolor='lightyellow'
)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Plot with Info Boxes')

plt.show()
```

## Complete Publication Figure

A multi-panel figure combining various annotation techniques using SubplotLayout.

```python
from yplot.annotate.elements import add_region_highlight

# Create a 2-panel publication figure
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 4.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 3.5),
        "cols": 2,
        "spacing": {"hspace": 1.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Panel A: Bar chart with significance
ax = axes[0]
bars = ax.bar(range(3), [10, 25, 18], yerr=[2, 4, 3], capsize=3)
ax.set_xticks(range(3))
ax.set_xticklabels(['Control', 'Treatment', 'Recovery'])
ax.set_ylabel('Response')
ax.set_ylim(0, 35)

add_significance(ax, 0, 1, 30, pvalue=0.001)
add_text_box(ax, 0.02, 0.98, 'A', fontsize=12, transform='axes', 
             facecolor='none', edgecolor='none')

# Panel B: Time course with region highlight
ax = axes[1]
t = np.linspace(0, 20, 100)
y = np.exp(-0.1 * t) * np.sin(t)
ax.plot(t, y, 'b-', lw=1.5)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Signal')

# Highlight treatment period
add_region_highlight(ax, 5, 10, color='yellow', alpha=0.3, label='Treatment')
add_text_box(ax, 0.02, 0.98, 'B', fontsize=12, transform='axes',
             facecolor='none', edgecolor='none')
ax.legend(loc='upper right')

plt.show()
```

## Multi-Row Annotation Figure

```python
np.random.seed(42)

# Create a 2x2 layout for comprehensive annotation examples
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 8),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    },
    "row_2": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Panel A: Bar chart with significance
ax = axes[0]
ax.bar(range(3), [15, 28, 22], yerr=[2, 3, 2.5], capsize=3)
ax.set_xticks(range(3))
ax.set_xticklabels(['Ctrl', 'Trt', 'Rec'])
ax.set_ylabel('Value')
ax.set_ylim(0, 40)
add_significance(ax, 0, 1, 33, pvalue=0.001)
ax.set_title('Significance Brackets')

# Panel B: Regression with stats
ax = axes[1]
x = np.linspace(0, 10, 30)
y = 2 * x + 5 + np.random.normal(0, 2, 30)
ax.scatter(x, y, alpha=0.7)
slope, intercept, r, p, _ = stats.linregress(x, y)
ax.plot(x, slope * x + intercept, 'r-', lw=1.5)
add_regression_stats(ax, r**2, p, slope, intercept, loc='lower right')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Regression Statistics')

# Panel C: Callouts and arrows
ax = axes[2]
t = np.linspace(0, 4*np.pi, 100)
y = np.sin(t)
ax.plot(t, y)
peak_idx = np.argmax(y)
add_callout(ax, t[peak_idx], y[peak_idx], 'Peak', t[peak_idx]+1.5, y[peak_idx]+0.2)
ax.set_xlabel('Time')
ax.set_ylabel('Amplitude')
ax.set_title('Callouts')

# Panel D: Text boxes
ax = axes[3]
x = np.linspace(0, 5, 50)
y = np.exp(x/2)
ax.plot(x, y)
add_text_box(ax, 0.05, 0.95, 'n = 50', transform='axes')
add_text_box(ax, 0.95, 0.3, 'Exponential\ngrowth', transform='axes', ha='right', facecolor='lightyellow')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Text Boxes')

# Add subplot labels
yplot.add_subplot_labels(fig, axes)

plt.show()
```
