# Heatmaps

This notebook demonstrates yplot's heatmap functionality for:
- Correlation matrices
- Data matrices
- Annotated heatmaps

**All examples use SubplotLayout** for precise control over figure dimensions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import yplot
import yplot.plot as yp

yplot.use('publication')
```

## Basic Heatmap from Array

```python
# Create sample data
np.random.seed(42)
data = np.random.randn(5, 8)

layout = yplot.SubplotLayout(config={
    "fig_size": (8, 4.5),
    "margins": {"left": 0.5, "right": 0.8, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (6.5, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
yp.heatmap(data, cmap='viridis', ax=axes[0])
axes[0].set_title('Basic Heatmap')

plt.show()
```

## Heatmap from DataFrame

```python
# Create labeled DataFrame
df = pd.DataFrame(
    np.random.randn(4, 6),
    index=['Gene A', 'Gene B', 'Gene C', 'Gene D'],
    columns=['Sample 1', 'Sample 2', 'Sample 3', 'Sample 4', 'Sample 5', 'Sample 6']
)

layout = yplot.SubplotLayout(config={
    "fig_size": (8, 4.5),
    "margins": {"left": 0.8, "right": 0.8, "top": 0.4, "bottom": 0.6},
    "row_1": {
        "size": (6.0, 3.2),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
yp.heatmap(df, cmap='RdBu_r', center=0, ax=axes[0])
axes[0].set_title('Gene Expression Heatmap')

plt.show()
```

## Annotated Heatmap

```python
# Smaller matrix for annotations
small_df = pd.DataFrame(
    np.random.randn(4, 4),
    index=['A', 'B', 'C', 'D'],
    columns=['W', 'X', 'Y', 'Z']
)

layout = yplot.SubplotLayout(config={
    "fig_size": (5.5, 4.5),
    "margins": {"left": 0.5, "right": 0.8, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
yp.heatmap(small_df, annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=axes[0])
axes[0].set_title('Annotated Heatmap')

plt.show()
```

## Correlation Matrix

```python
# Create correlated data
np.random.seed(42)
n = 100
data = pd.DataFrame({
    'Var A': np.random.randn(n),
    'Var B': np.random.randn(n),
    'Var C': np.random.randn(n),
    'Var D': np.random.randn(n),
})
# Add correlations
data['Var B'] = data['Var A'] * 0.8 + data['Var B'] * 0.2
data['Var D'] = data['Var C'] * -0.5 + data['Var D'] * 0.5

# Compute correlation
corr = data.corr()

layout = yplot.SubplotLayout(config={
    "fig_size": (5.5, 4.5),
    "margins": {"left": 0.7, "right": 0.8, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.8, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
yp.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True, ax=axes[0])
axes[0].set_title('Correlation Matrix')

plt.show()
```

## Heatmap from Long-Form Data

```python
# Long-form data (common from experiments)
long_df = pd.DataFrame({
    'row': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
    'col': ['X', 'Y', 'Z', 'X', 'Y', 'Z', 'X', 'Y', 'Z'],
    'value': [1, 2, 3, 4, 5, 6, 7, 8, 9],
})

layout = yplot.SubplotLayout(config={
    "fig_size": (4.5, 3.5),
    "margins": {"left": 0.5, "right": 0.6, "top": 0.4, "bottom": 0.4},
    "row_1": {
        "size": (3.2, 2.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
yp.heatmap(long_df, x='col', y='row', values='value', annot=True, fmt='.0f', ax=axes[0])
axes[0].set_title('Heatmap from Long-Form Data')

plt.show()
```

## Comparing Different Colormaps

```python
data = np.random.randn(4, 5)

layout = yplot.SubplotLayout(config={
    "fig_size": (12, 4),
    "margins": {"left": 0.3, "right": 0.6, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.2, 2.8),
        "cols": 3,
        "spacing": {"hspace": 0.8, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Different colormaps
yp.heatmap(data, cmap='viridis', ax=axes[0])
axes[0].set_title('Viridis')

yp.heatmap(data, cmap='plasma', ax=axes[1])
axes[1].set_title('Plasma')

yp.heatmap(data, cmap='RdBu_r', center=0, ax=axes[2])
axes[2].set_title('Diverging (centered at 0)')

plt.show()
```

## Multi-Panel Heatmap Figure

Creating a publication-ready figure with multiple heatmaps using SubplotLayout.

```python
# Generate different datasets
np.random.seed(42)

# Gene expression data
genes = pd.DataFrame(
    np.random.randn(6, 4),
    index=[f'Gene {i}' for i in range(1, 7)],
    columns=['Ctrl', 'Drug A', 'Drug B', 'Drug C']
)

# Correlation matrix
corr_data = pd.DataFrame(np.random.randn(50, 4), columns=['A', 'B', 'C', 'D'])
corr_data['B'] = corr_data['A'] * 0.8 + corr_data['B'] * 0.2
corr = corr_data.corr()

# 2-row layout
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 8),
    "margins": {"left": 0.8, "right": 0.8, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 1.2, "wspace": 0.6}
    },
    "row_2": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 1.2, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Gene expression heatmap
yp.heatmap(genes, cmap='RdBu_r', center=0, ax=axes[0])
axes[0].set_title('Gene Expression')

# Correlation matrix
yp.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True, ax=axes[1])
axes[1].set_title('Correlation Matrix')

# Annotated small heatmap
small_data = np.random.randn(3, 3)
yp.heatmap(small_data, annot=True, fmt='.2f', cmap='viridis', ax=axes[2])
axes[2].set_title('Annotated Values')

# Large matrix without annotations
large_data = np.random.randn(8, 6)
yp.heatmap(large_data, cmap='plasma', ax=axes[3])
axes[3].set_title('Large Matrix')

# Add subplot labels
yplot.add_subplot_labels(fig, axes)

plt.show()
```
