# SubplotLayout System

This notebook demonstrates yplot's **SubplotLayout** system, the core feature for creating publication-quality figures with precise control over:

- Figure dimensions in inches
- Subplot sizes in inches
- Margins and spacing
- Multi-row layouts with different column counts
- YAML configuration support

**Why use SubplotLayout instead of plt.subplots()?**
- Exact control over subplot dimensions (e.g., 2.5" x 2.5" panels)
- Different row heights and column counts per row
- Explicit spacing in inches rather than relative units
- Better for journal-specific figure requirements

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import yplot
import yplot.plot as yp

yplot.use('publication')
```

## Basic Single Subplot

Even for a single subplot, SubplotLayout gives you precise control.

```python
# Define a single 4x3 inch subplot
layout = yplot.SubplotLayout(config={
    "fig_size": (5, 4),           # Total figure size
    "margins": {
        "left": 0.6,               # Space for y-axis label
        "right": 0.2,              # Right padding
        "top": 0.3,                # Space for title
        "bottom": 0.5              # Space for x-axis label
    },
    "row_1": {
        "size": (4.0, 3.0),        # Subplot size in inches
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Plot some data
x = np.linspace(0, 10, 50)
axes[0].plot(x, np.sin(x), 'b-', lw=1.5)
axes[0].set_xlabel('X')
axes[0].set_ylabel('Y')
axes[0].set_title('Single Subplot (4x3 inches)')

plt.show()
```

## Single Row with Multiple Columns

Create a row of equal-sized subplots with controlled spacing.

```python
# Three 2.5x2.5 inch subplots in a row
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 3.5),
    "margins": {"left": 0.5, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (2.5, 2.5),        # Each subplot is 2.5x2.5 inches
        "cols": 3,                  # 3 columns
        "spacing": {
            "hspace": 0.7,          # Horizontal space between columns
            "wspace": 0.0           # Vertical space (not used for single row)
        }
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Different plots in each panel
x = np.linspace(0, 2*np.pi, 50)
axes[0].plot(x, np.sin(x))
axes[0].set_title('Sine')

axes[1].plot(x, np.cos(x))
axes[1].set_title('Cosine')

axes[2].plot(x, np.sin(x) * np.cos(x))
axes[2].set_title('Sin * Cos')

for ax in axes:
    ax.set_xlabel('X')
    ax.set_ylabel('Y')

plt.show()
```

## Multiple Rows with Same Column Count

A classic 2x2 grid layout.

```python
# 2x2 grid with 3x2.5 inch subplots
layout = yplot.SubplotLayout(config={
    "fig_size": (8, 7),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.7, "wspace": 0.6}  # wspace = vertical between rows
    },
    "row_2": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.7, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Sample data
np.random.seed(42)
x = np.linspace(0, 10, 50)

axes[0].scatter(x, x + np.random.normal(0, 1, 50), alpha=0.7)
axes[0].set_title('Scatter Plot')

axes[1].plot(x, np.exp(x/5))
axes[1].set_title('Exponential')

axes[2].bar([1, 2, 3, 4], [10, 25, 15, 30])
axes[2].set_title('Bar Chart')

axes[3].hist(np.random.randn(500), bins=20)
axes[3].set_title('Histogram')

# Add subplot labels (A, B, C, D)
yplot.add_subplot_labels(fig, axes)

plt.show()
```

## Different Column Counts Per Row

This is where SubplotLayout really shines - different numbers of columns in each row.

```python
# Row 1: 3 small panels, Row 2: 2 larger panels
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 7),
    "margins": {"left": 0.5, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (2.5, 2.5),        # Smaller panels
        "cols": 3,
        "spacing": {"hspace": 0.6, "wspace": 0.6}
    },
    "row_2": {
        "size": (4.0, 2.5),        # Larger panels
        "cols": 2,
        "spacing": {"hspace": 0.6, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Top row: 3 scatter plots
np.random.seed(42)
for i in range(3):
    x = np.random.randn(30)
    y = np.random.randn(30)
    axes[i].scatter(x, y, alpha=0.7)
    axes[i].set_title(f'Scatter {i+1}')
    axes[i].set_xlabel('X')
    axes[i].set_ylabel('Y')

# Bottom row: 2 line plots
x = np.linspace(0, 10, 100)
axes[3].plot(x, np.sin(x), 'b-', label='sin(x)')
axes[3].plot(x, np.cos(x), 'r--', label='cos(x)')
axes[3].legend()
axes[3].set_title('Trigonometric Functions')
axes[3].set_xlabel('X')
axes[3].set_ylabel('Y')

axes[4].fill_between(x, np.sin(x) - 0.2, np.sin(x) + 0.2, alpha=0.3)
axes[4].plot(x, np.sin(x), 'b-')
axes[4].set_title('Line with Error Band')
axes[4].set_xlabel('X')
axes[4].set_ylabel('Y')

# Add subplot labels
yplot.add_subplot_labels(fig, axes)

plt.show()
```

## Different Row Heights

Rows can have different heights for specialized layouts.

```python
# Top row: tall panels, Bottom row: short wide panels
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 8),
    "margins": {"left": 0.5, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 4.0),        # Tall panels
        "cols": 2,
        "spacing": {"hspace": 1.0, "wspace": 0.8}
    },
    "row_2": {
        "size": (4.0, 2.0),        # Short wide panels
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.8}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

np.random.seed(42)

# Top row: Tall panels for vertical data
axes[0].barh([1, 2, 3, 4, 5], [10, 25, 15, 30, 20])
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Category')
axes[0].set_title('Horizontal Bars')

y = np.linspace(0, 10, 50)
axes[1].plot(np.sin(y), y)
axes[1].set_xlabel('X')
axes[1].set_ylabel('Y')
axes[1].set_title('Vertical Line')

# Bottom row: Wide panels for time series
t = np.linspace(0, 20, 200)
axes[2].plot(t, np.sin(t) * np.exp(-t/10))
axes[2].set_xlabel('Time')
axes[2].set_ylabel('Signal')
axes[2].set_title('Damped Oscillation')

axes[3].plot(t, np.random.randn(200).cumsum())
axes[3].set_xlabel('Time')
axes[3].set_ylabel('Value')
axes[3].set_title('Random Walk')

yplot.add_subplot_labels(fig, axes)

plt.show()
```

## Three-Row Layout

Complex multi-row layouts for comprehensive figures.

```python
# 3-row layout: 1 column, 3 columns, 2 columns
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 10),
    "margins": {"left": 0.5, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (8.5, 2.5),        # Wide single panel
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.6}
    },
    "row_2": {
        "size": (2.5, 2.5),        # 3 small panels
        "cols": 3,
        "spacing": {"hspace": 0.6, "wspace": 0.6}
    },
    "row_3": {
        "size": (4.0, 2.5),        # 2 medium panels
        "cols": 2,
        "spacing": {"hspace": 0.6, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

np.random.seed(42)

# Row 1: Wide overview panel
t = np.linspace(0, 50, 500)
axes[0].plot(t, np.sin(t/5) + 0.5*np.sin(t) + np.random.normal(0, 0.1, 500))
axes[0].set_xlabel('Time')
axes[0].set_ylabel('Signal')
axes[0].set_title('Overview Time Series')

# Row 2: Detail panels
for i in range(3):
    ax = axes[1 + i]
    x = np.random.randn(50)
    ax.hist(x, bins=15, alpha=0.7)
    ax.set_title(f'Distribution {i+1}')
    ax.set_xlabel('Value')
    ax.set_ylabel('Count')

# Row 3: Summary panels
ax = axes[4]
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 12, 38]
ax.bar(categories, values)
ax.set_title('Summary Statistics')
ax.set_xlabel('Category')
ax.set_ylabel('Value')

ax = axes[5]
x = np.linspace(0, 10, 50)
y = 2*x + 3 + np.random.normal(0, 2, 50)
ax.scatter(x, y, alpha=0.7)
ax.plot(x, 2*x + 3, 'r-', lw=2)
ax.set_title('Correlation')
ax.set_xlabel('X')
ax.set_ylabel('Y')

yplot.add_subplot_labels(fig, axes)

plt.show()
```

## Understanding Spacing

Visualizing how `hspace` (horizontal) and `wspace` (vertical) affect layout.

```python
# Tight spacing
layout_tight = yplot.SubplotLayout(config={
    "fig_size": (8, 7),
    "margins": {"left": 0.5, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.3, "wspace": 0.3}  # Tight
    },
    "row_2": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.3, "wspace": 0.3}
    }
})

fig, axes = yplot.create_figure_with_layout(layout_tight)
fig.suptitle('Tight Spacing (0.3 inch)', y=0.98)

for i, ax in enumerate(axes):
    ax.plot([0, 1], [0, 1])
    ax.set_title(f'Panel {i+1}')

plt.show()
```

```python
# Wide spacing
layout_wide = yplot.SubplotLayout(config={
    "fig_size": (9, 8),
    "margins": {"left": 0.5, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 1.0, "wspace": 1.0}  # Wide
    },
    "row_2": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 1.0, "wspace": 1.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout_wide)
fig.suptitle('Wide Spacing (1.0 inch)', y=0.98)

for i, ax in enumerate(axes):
    ax.plot([0, 1], [0, 1])
    ax.set_title(f'Panel {i+1}')

plt.show()
```

## Saving Layout to YAML

Layouts can be saved to and loaded from YAML files for reuse.

```python
# Define a layout
layout_config = {
    "fig_size": (10, 8),
    "margins": {"left": 0.6, "right": 0.3, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    },
    "row_2": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    }
}

layout = yplot.SubplotLayout(config=layout_config)

# Save to YAML (commented out to avoid file creation)
# layout.to_yaml('my_layout.yaml')

# Load from YAML
# layout = yplot.SubplotLayout(yaml_file='my_layout.yaml')

# Print the configuration
print("Layout Configuration:")
print(f"  Figure size: {layout.fig_size_inches}")
print(f"  Number of rows: {layout.rows}")
print(f"  Margins: {layout.margins}")
```

## Working with Coordinates

Understanding how SubplotLayout calculates subplot positions.

```python
# Simple 2x2 layout
layout = yplot.SubplotLayout(config={
    "fig_size": (8, 6),
    "margins": {"left": 0.5, "right": 0.5, "top": 0.5, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 2.0),
        "cols": 2,
        "spacing": {"hspace": 0.5, "wspace": 0.5}
    },
    "row_2": {
        "size": (3.0, 2.0),
        "cols": 2,
        "spacing": {"hspace": 0.5, "wspace": 0.5}
    }
})

# Get coordinates (left, bottom, width, height) in figure-relative units (0-1)
coords = layout.get_coordinates()

print("Subplot coordinates (left, bottom, width, height):")
for i, coord in enumerate(coords):
    print(f"  Subplot {i}: ({coord[0]:.3f}, {coord[1]:.3f}, {coord[2]:.3f}, {coord[3]:.3f})")
```

## rcParams Integration

SubplotLayout works with yplot's configuration system.

```python
# Use rc_context to temporarily change settings
with yplot.rc_context({'font.size': 10, 'axes.linewidth': 1.5}):
    layout = yplot.SubplotLayout(config={
        "fig_size": (8, 4),
        "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
        "row_1": {
            "size": (3.0, 3.0),
            "cols": 2,
            "spacing": {"hspace": 0.8, "wspace": 0.0}
        }
    })
    
    fig, axes = yplot.create_figure_with_layout(layout)
    
    x = np.linspace(0, 10, 50)
    axes[0].plot(x, np.sin(x))
    axes[0].set_title('Larger Font Size')
    axes[0].set_xlabel('X')
    axes[0].set_ylabel('Y')
    
    axes[1].plot(x, np.cos(x))
    axes[1].set_title('Thicker Lines')
    axes[1].set_xlabel('X')
    axes[1].set_ylabel('Y')
    
    plt.show()

print("Settings restored to defaults outside context.")
```
