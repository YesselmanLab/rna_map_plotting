# Distribution Plots

This notebook demonstrates yplot's distribution visualization functions:
- Violin plots
- Box plots
- Swarm plots
- Strip plots

**All examples use SubplotLayout** for precise control over figure dimensions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import yplot
import yplot.plot as yp

yplot.use('publication')
```

## Create Sample Distribution Data

```python
np.random.seed(42)

# Create data with different distributions
df = pd.DataFrame({
    'treatment': (['Control'] * 50 + ['Drug A'] * 50 + 
                  ['Drug B'] * 50 + ['Drug C'] * 50),
    'response': np.concatenate([
        np.random.normal(10, 2, 50),    # Control
        np.random.normal(15, 3, 50),    # Drug A
        np.random.normal(12, 4, 50),    # Drug B
        np.random.normal(20, 2.5, 50),  # Drug C
    ])
})

df.head()
```

## Violin Plots

Violin plots show the full distribution shape.

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (6, 4.5),
    "margins": {"left": 0.7, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (5.0, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

yp.violin(df, x='treatment', y='response', ax=ax)
ax.set_ylabel('Response Value')
ax.set_title('Violin Plot - Treatment Comparison')

plt.show()
```

## Box Plots

Box plots show quartiles and outliers.

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (9, 4.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.5, 3.5),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Standard box plot
yp.box(df, x='treatment', y='response', ax=axes[0])
axes[0].set_title('Standard Box Plot')

# Notched box plot
yp.box(df, x='treatment', y='response', notch=True, ax=axes[1])
axes[1].set_title('Notched Box Plot')

plt.show()
```

## Swarm and Strip Plots

These show individual data points.

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (9, 4.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.5, 3.5),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Swarm plot - points spread to avoid overlap
yp.swarm(df, x='treatment', y='response', ax=axes[0])
axes[0].set_title('Swarm Plot')

# Strip plot - random jitter
yp.strip(df, x='treatment', y='response', jitter=0.2, ax=axes[1])
axes[1].set_title('Strip Plot')

plt.show()
```

## Combining Distribution Plots

A common pattern is to overlay a violin/box with individual points.

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (6, 4.5),
    "margins": {"left": 0.7, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (5.0, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

# Draw violin first
yp.violin(df, x='treatment', y='response', alpha=0.3, ax=ax)

# Overlay strip plot
yp.strip(df, x='treatment', y='response', size=3, alpha=0.7, ax=ax)

ax.set_title('Violin + Strip Overlay')

plt.show()
```

## Box Plot with Individual Points

```python
layout = yplot.SubplotLayout(config={
    "fig_size": (6, 4.5),
    "margins": {"left": 0.7, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (5.0, 3.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
ax = axes[0]

# Draw box without fliers (we'll show all points)
yp.box(df, x='treatment', y='response', showfliers=False, ax=ax)

# Overlay all individual points
yp.swarm(df, x='treatment', y='response', size=4, alpha=0.6, ax=ax)

ax.set_title('Box Plot with Data Points')

plt.show()
```

## All Distribution Types Comparison

Side-by-side comparison of all distribution visualization methods.

```python
# Create a 2x2 layout for comparing all distribution types
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 8),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    },
    "row_2": {
        "size": (4.0, 3.0),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Violin
yp.violin(df, x='treatment', y='response', ax=axes[0])
axes[0].set_title('Violin Plot')

# Box
yp.box(df, x='treatment', y='response', ax=axes[1])
axes[1].set_title('Box Plot')

# Swarm
yp.swarm(df, x='treatment', y='response', ax=axes[2])
axes[2].set_title('Swarm Plot')

# Strip
yp.strip(df, x='treatment', y='response', jitter=0.2, ax=axes[3])
axes[3].set_title('Strip Plot')

# Add subplot labels
yplot.add_subplot_labels(fig, axes)

plt.show()
```
