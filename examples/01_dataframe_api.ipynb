# DataFrame-First Plotting API

This notebook demonstrates yplot's new DataFrame-first plotting API, similar to seaborn.

The key advantage is that you pass column names instead of arrays, making code more readable.

**All examples use SubplotLayout** for precise control over figure dimensions and spacing.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import yplot
import yplot.plot as yp

# Apply publication style
yplot.use('publication')
```

## Create Sample Data

```python
# Create sample experimental data
np.random.seed(42)
n_samples = 50

df = pd.DataFrame({
    'concentration': np.linspace(0.1, 10, n_samples),
    'response': 2.5 * np.linspace(0.1, 10, n_samples) + np.random.normal(0, 2, n_samples),
    'treatment': np.random.choice(['Control', 'Drug A', 'Drug B'], n_samples),
    'replicate': np.random.choice(['Rep1', 'Rep2', 'Rep3'], n_samples),
})

df.head()
```

## Scatter Plots with SubplotLayout

Using SubplotLayout gives precise control over subplot sizes in inches.

```python
# Define layout: 3 columns, each 2.5 x 2.5 inches
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 3.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (2.5, 2.5),
        "cols": 3,
        "spacing": {"hspace": 0.8, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Basic scatter
yp.scatter(df, x='concentration', y='response', ax=axes[0])
axes[0].set_title('Basic Scatter')

# Scatter with color grouping
yp.scatter(df, x='concentration', y='response', color='treatment', ax=axes[1])
axes[1].set_title('Grouped by Treatment')

# Scatter with custom size
yp.scatter(df, x='concentration', y='response', size=100, alpha=0.5, ax=axes[2])
axes[2].set_title('Custom Size & Alpha')

plt.show()
```

## Line Plots with Error Bands

```python
# Create time series data
time_df = pd.DataFrame({
    'time': np.tile(np.arange(0, 10, 0.5), 3),
    'measurement': np.concatenate([
        np.sin(np.arange(0, 10, 0.5)) + np.random.normal(0, 0.1, 20),
        np.sin(np.arange(0, 10, 0.5)) * 1.5 + np.random.normal(0, 0.1, 20),
        np.sin(np.arange(0, 10, 0.5)) * 0.5 + np.random.normal(0, 0.1, 20),
    ]),
    'condition': ['A'] * 20 + ['B'] * 20 + ['C'] * 20,
})

# Add error column
time_df['error'] = 0.15

time_df.head()
```

```python
# Define layout: 2 columns
layout = yplot.SubplotLayout(config={
    "fig_size": (8, 3.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Single line with error band
subset = time_df[time_df['condition'] == 'A']
yp.line(subset, x='time', y='measurement', error='error', ax=axes[0])
axes[0].set_title('Line with Error Band')

# Multiple lines grouped by condition
yp.line(time_df, x='time', y='measurement', group='condition', ax=axes[1])
axes[1].set_title('Grouped Lines')

plt.show()
```

## Bar Plots

```python
# Create summary data
summary_df = pd.DataFrame({
    'condition': ['Control', 'Low Dose', 'High Dose'],
    'mean_response': [10, 25, 45],
    'std': [2, 5, 8],
})

# Grouped bar data
grouped_df = pd.DataFrame({
    'condition': ['Control', 'Control', 'Treatment', 'Treatment'],
    'time': ['Day 1', 'Day 7', 'Day 1', 'Day 7'],
    'value': [10, 15, 25, 45],
})
```

```python
# Define layout for bar plots
layout = yplot.SubplotLayout(config={
    "fig_size": (8, 3.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (3.0, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Basic bar with error
yp.bar(summary_df, x='condition', y='mean_response', error='std', ax=axes[0])
axes[0].set_title('Bar with Error Bars')

# Grouped bar
yp.grouped_bar(grouped_df, x='condition', y='value', group='time', ax=axes[1])
axes[1].set_title('Grouped Bar')

plt.show()
```

## Working with Arrays (Non-DataFrame)

You can also use the API with plain arrays when you don't have a DataFrame.

```python
x = np.linspace(0, 10, 50)
y = np.sin(x) + np.random.normal(0, 0.1, 50)

# Single subplot layout
layout = yplot.SubplotLayout(config={
    "fig_size": (5, 3.5),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.4, "bottom": 0.5},
    "row_1": {
        "size": (4.0, 2.5),
        "cols": 1,
        "spacing": {"hspace": 0.0, "wspace": 0.0}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)
yp.scatter(x=x, y=y, size=30, alpha=0.7, ax=axes[0])
axes[0].set_title('Array Input')
plt.show()
```

## Multi-Row Layout Example

SubplotLayout makes it easy to create complex multi-row figures.

```python
# Define a 2-row layout: 3 columns on top, 2 columns on bottom
layout = yplot.SubplotLayout(config={
    "fig_size": (10, 7),
    "margins": {"left": 0.6, "right": 0.2, "top": 0.3, "bottom": 0.5},
    "row_1": {
        "size": (2.5, 2.5),
        "cols": 3,
        "spacing": {"hspace": 0.7, "wspace": 0.6}
    },
    "row_2": {
        "size": (3.5, 2.5),
        "cols": 2,
        "spacing": {"hspace": 0.8, "wspace": 0.6}
    }
})

fig, axes = yplot.create_figure_with_layout(layout)

# Top row: Three scatter plots
yp.scatter(df, x='concentration', y='response', ax=axes[0])
axes[0].set_title('Scatter 1')

yp.scatter(df, x='concentration', y='response', color='treatment', ax=axes[1])
axes[1].set_title('Scatter 2')

yp.scatter(df, x='concentration', y='response', size=80, alpha=0.5, ax=axes[2])
axes[2].set_title('Scatter 3')

# Bottom row: Line and bar plots
yp.line(time_df, x='time', y='measurement', group='condition', ax=axes[3])
axes[3].set_title('Grouped Lines')

yp.grouped_bar(grouped_df, x='condition', y='value', group='time', ax=axes[4])
axes[4].set_title('Grouped Bar')

# Add subplot labels (A, B, C, ...)
yplot.add_subplot_labels(fig, layout.get_final_coordinates())

plt.show()
```
